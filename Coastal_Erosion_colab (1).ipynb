```python
data.drop(columns=["Annual balance"], inplace=True)
# data.drop(['Annual balance'],axis=1)
print(data.columns)
```

```python
data_cleaned = data.dropna(subset=['Land Impact', 'Year', 'Country', 'Continent', 'Land area', 'Coastal length', 'Temperature', 'Sea level'])
ip = data_cleaned[['Year', 'Country', 'Continent', 'Land area', 'Coastal length', 'Temperature', 'Sea level']]
op = data_cleaned['Land Impact']
```

```python
from sklearn.model_selection import train_test_split
xtr,xts,ytr,yts=train_test_split(ip,op,test_size=0.2)

print(f"Training features shape: {xtr.shape}")
print(f"Testing features shape: {xts.shape}")
print(f"Training target shape: {ytr.shape}")
print(f"Testing target shape: {yts.shape}")
```

```python
print(ip.isnull().sum())
print(op.isnull().sum())
```

```python
from sklearn.preprocessing import StandardScaler
sc=StandardScaler()
sc.fit(xtr)
sc.fit(xts)
xtr=sc.transform(xtr)
xts=sc.transform(xts)
```

```python
from sklearn.linear_model import LinearRegression
import time
from sklearn.metrics import mean_squared_error, r2_score

model = LinearRegression()

start_time = time.time()
model.fit(xtr, ytr)
Y_pred = model.predict(xts)

mse = mean_squared_error(yts, Y_pred)
r2 = r2_score(yts, Y_pred)
exec_time = time.time() - start_time

print(f"Linear Regression -> MSE: {mse:.4f}, R² Score: {r2:.4f}, Execution Time: {exec_time:.4f} sec")
```

```python
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression

poly = PolynomialFeatures(degree=2)
X_train_poly = poly.fit_transform(xtr)
X_test_poly = poly.transform(xts)

model = LinearRegression()

start_time = time.time()
model.fit(X_train_poly, ytr)
Y_pred = model.predict(X_test_poly)

mse = mean_squared_error(yts, Y_pred)
r2 = r2_score(yts, Y_pred)
exec_time = time.time() - start_time

print(f"Polynomial Regression -> MSE: {mse:.4f}, R² Score: {r2:.4f}, Execution Time: {exec_time:.4f} sec")
```

```python
from sklearn.ensemble import StackingRegressor

stacking_model = StackingRegressor(
    estimators=[
        ("LR", LinearRegression()),
        ("RF", RandomForestRegressor(n_estimators=50)),
        ("XGB", XGBRegressor(objective='reg:squarederror'))
    ],
    final_estimator=RandomForestRegressor(n_estimators=100)
)

start_time = time.time()
stacking_model.fit(xtr, ytr)
Y_pred = stacking_model.predict(xts)

mse = mean_squared_error(yts, Y_pred)
r2 = r2_score(yts, Y_pred)
exec_time = time.time() - start_time

print(f"Stacking Regressor -> MSE: {mse:.4f}, R² Score: {r2:.4f}, Execution Time: {exec_time:.4f} sec")
```

```python
import numpy as np
import time
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error, r2_score

model = RandomForestRegressor(n_estimators=100)

start_time = time.time()
model.fit(xtr, ytr)
Y_pred = model.predict(xts)

mse = mean_squared_error(yts, Y_pred)
r2 = r2_score(yts, Y_pred)
exec_time = time.time() - start_time

print(f"Random Forest -> MSE: {mse:.4f}, R² Score: {r2:.4f}, Execution Time: {exec_time:.4f} sec")

def predict_land_impact(model, year, country, continent, land_area, coastal_length, temperature, sea_level, scaler=None):
    input_data = np.array([[year, country, continent, land_area, coastal_length, temperature, sea_level]])

    if scaler:
        input_data = scaler.transform(input_data)

    predicted_land_impact = model.predict(input_data)
    return predicted_land_impact[0]


example_prediction = predict_land_impact(model,
                                         year=2025,
                                         country=10,
                                         continent=2,
                                         land_area=500000,
                                         coastal_length=1000,
                                         temperature=15.5,
                                         sea_level=3.2,
                                         scaler=sc)

print(f"\nPredicted Land Impact: {example_prediction:.4f}")
```
